# Variables and Data Types
### Variables are used to store data in memory so that it can be reused later.
✅ Key Points to Remember

->No need to define data type

->Variable values can be changed

->Python is dynamically typed

```python
name = "Abhishek"
age = 20
cgpa = 8.5

print("Name:", name)
print("Age:", age)
print("CGPA:", cgpa)
```

```python
age = 21 
print("Updated Age:", age)
cgpa = 9.34
print("Updated CGPA:", cgpa)
name = "Abhishek Kumar"
print("Updated Name:", name)
passed = True
print(passed)
x = 10
print(type(x))      # <class 'int'>
a, b, c = 1, 2, 3
print(a, b, c)
```

```python
print(3)
print(358)
print(50000)
print(3 + 3)
print(2 * 5)
# Mix Text and Numbers
print("I am", 35, "years old.")

```

```python
x = "Python "
y = "is "
z = "awesome"
print(x + y + z)
# OR
x = "Python"
y = "is"
z = "awesome"
print(x, y, z)
```

```python
# GLOBAL VARIABLE
def myfunc():
  global x
  x = "fantastic"

myfunc()

print("Python is " + x)
```

```python
x = "Hello World"	#str	
x = 20	#int	
x = 20.5	#float	
x = 1j	#complex	
x = ["apple", "banana", "cherry"]	#list	
x = ("apple", "banana", "cherry")	#tuple	
x = range(6)	#range	
x = {"name" : "John", "age" : 36}	#dict
x = {"apple", "banana", "cherry"}	#set	
x = frozenset({"apple", "banana", "cherry"})	#frozenset	
x = True	#bool	
x = b"Hello"	#bytes	
x = bytearray(5)	#bytearray
x = memoryview(bytes(5))	#memoryview	
x = None
```

```python
x = int(1)   # x will be 1
y = float(2.8)   # y will be 2.8
y = str(2)    # y will be '2'
```

# Conditions and If statements

```python
# Equals: a == b
# Not Equals: a != b
# Less than: a < b
# Less than or equal to: a <= b
# Greater than: a > b
# Greater than or equal to: a >= b

a = 200
b = 33
if b > a:
  print("b is greater than a")
elif a == b:
  print("a and b are equal")
else:
  print("a is greater than b")
```

```python
# One-line if statement:
a = 5
b = 2
if a > b: print("a is greater than b")
```

```python
age = 25
is_student = False
has_discount_code = True

if (age < 18 or age > 65) and not is_student or has_discount_code:
  print("Discount applies!")
```

```python
# NESTED IF STATEMENTS
x = 41

if x > 10:
  print("Above ten,")
  if x > 20:
    print("and also above 20!")
  else:
    print("but not above 20.")
```

### PASS
Why Use pass?
The pass statement is useful in several situations:

When you're creating code structure but haven't implemented the logic yet
When a statement is required syntactically but no action is needed
As a placeholder for future code during development
In empty functions or classes that you plan to implement later

```python
age = 16

if age < 18:
  pass # TODO: Add underage logic later
else:
  print("Access granted")
```

# MATCH statement

```python
day = 4
match day:
  case 1:
    print("Monday")
  case 2:
    print("Tuesday")
  case 3:
    print("Wednesday")
  case 4:
    print("Thursday")
  case 5:
    print("Friday")
  case 6:
    print("Saturday")
  case 7:
    print("Sunday")
```

Default Value


```python
day = 4
match day:
  case 6:
    print("Today is Saturday")
  case 7:
    print("Today is Sunday")
    # THIS is the default value 
  case _:
    print("Looking forward to the Weekend")
    
```

```python
#COMBINE VALUE
day = 4
match day:
  case 1 | 2 | 3 | 4 | 5:
    print("Today is a weekday")
  case 6 | 7:
    print("I love weekends!")
```

#  Loops

```python
# The while Loop
i = 1
while i < 6:
  print(i)
  i += 1
```

```python
# The break Statement
i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i += 1
```

```python
#The continue Statement
i = 0
while i < 6:
  i += 1
  if i == 3:
    continue
  print(i)
```

```python
# else Statement
i = 1
while i < 6:
  print(i)
  i += 1
else:
  print("i is no longer less than 6")
```

```python
# For Loops
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
```

```python
for x in "banana":
  print(x)
```

```python
for x in range(6):
  print(x)
```

```python
for x in range(2, 30, 3):
  print(x)
```

```python
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
  for y in fruits:
    print(x, y)
```

```python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x)
  if x == "banana":
    break
```
